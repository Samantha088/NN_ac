# **Step 2 — Import Libraries**

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import matplotlib.pyplot as plt

```

# **Step 3 — Load Dataset**

```python
import random

positive_comments = [
    "I love this video!", "Amazing content!", "This helps a lot", "Great explanation",
    "Very useful tutorial", "Awesome editing!", "You earned a subscriber", "Well done!",
    "This is inspiring", "Perfectly explained", "I enjoyed watching this!", "So helpful!","Good day"
]

negative_comments = [
    "I hate this video", "Worst content ever", "This is so confusing",
    "Poor audio quality", "Waste of time", "Disliked this", "Very boring",
    "Not helpful at all", "Terrible explanation", "Bad editing","it's not bad"
]

neutral_comments = [
    "Nice video", "Okay content", "Interesting topic", "I expected more",
    "Not bad", "Could be better", "It's fine", "Average video"
]

data = []
for _ in range(200):
    sentiment_type = random.choice(["positive", "negative", "neutral"])

    if sentiment_type == "positive":
        text = random.choice(positive_comments)
    elif sentiment_type == "negative":
        text = random.choice(negative_comments)
    else:
        text = random.choice(neutral_comments)

    data.append([text, sentiment_type])

df = pd.DataFrame(data, columns=["text", "sentiment"])
df.head()
```

# **Step 4 — Encode Sentiment Labels**

```python
le = LabelEncoder()
df['sentiment_encoded'] = le.fit_transform(df['sentiment'])
df.head()

```

# **Step 5 — Text Tokenization and Padding**

```python
texts = df['text'].astype(str).values
labels = df['sentiment_encoded'].values

tokenizer = Tokenizer(num_words=5000, oov_token="<OOV>")
tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)
padded = pad_sequences(sequences, maxlen=20, padding='post')

```

# **Step 6 — Train-Test Split**

```python
X_train, X_test, y_train, y_test = train_test_split(padded, labels, test_size=0.2, random_state=42)

print("Training Size:", X_train.shape)
print("Testing Size:", X_test.shape)
```

# **Step 7 — Build RNN (LSTM) Model**

```python
model = Sequential()
model.add(Embedding(5000, 64, input_length=20))
model.add(LSTM(64))
model.add(Dense(3, activation='softmax'))  # 3 classes

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

model.summary()

```

# **Step 8 — Train the Model**

```python
history = model.fit(X_train, y_train, epochs=10, batch_size=16, validation_split=0.2)

```

# **Step 9 — Evaluate the Model**

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Model Accuracy: {accuracy*100:.2f}%")
```

# **Step 10 — Test with Custom Input**

```python
def predict_sentiment(text):
    seq = tokenizer.texts_to_sequences([text])
    pad = pad_sequences(seq, maxlen=20)
    result = model.predict(pad)[0]
    sentiment = le.inverse_transform([np.argmax(result)])
    return sentiment[0]

print(predict_sentiment("This video was absolutely amazing!"))
print(predict_sentiment("Worst thing I have seen"))
print(predict_sentiment("It's ok, nothing special"))

```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.title("Training Accuracy Curve")
plt.show()
```
